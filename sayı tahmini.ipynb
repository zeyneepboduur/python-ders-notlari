```python
gizli_sayi = 7

deneme_sayisi = 0

while True:
    tahmin = int(input("lütfen tahmininizi giriniz:"))
    deneme_sayisi += 1

    if tahmin < gizli_sayi:
        print("Hata! Daha büyük bir tahmin yapınız.")
    elif tahmin > gizli_sayi:
        print("Hata! Daha küçük bir tahmin yapınız.")
    else:
        print(f"Tebrikler! {deneme_sayisi} denemede doğru tahmin ettiniz.")
        break



```
